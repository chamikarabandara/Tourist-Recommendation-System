```python
# 1. Create normalized district keys
accommodation["District_Key"] = (
    accommodation["District"]
    .astype("string")
    .str.strip()
    .str.casefold()
)

destination_master["District_Key"] = (
    destination_master["Mapped_District"]
    .astype("string")
    .str.strip()
    .str.casefold()
)

# 2. Count accommodations in each district
accommodation_counts = (
    accommodation
    .dropna(subset=["District_Key"])
    .groupby("District_Key")
    .size()
    .reset_index(name="Accommodation_Count")
)

# 3. Connect accommodation counts to destinations
destination_master = destination_master.drop(
    columns=["Accommodation_Count"],
    errors="ignore"
)

destination_master = destination_master.merge(
    accommodation_counts,
    on="District_Key",
    how="left",
    validate="many_to_one"
)

destination_master["Accommodation_Count"] = (
    destination_master["Accommodation_Count"]
    .fillna(0)
    .astype(int)
)

# 4. Validate results
print("Destination master shape:", destination_master.shape)
print("Unique destinations:",
      destination_master["Location_Name"].nunique())

print("\nDestinations with district-level accommodation matches:",
      (destination_master["Accommodation_Count"] > 0).sum())

print("\nDestinations without accommodation matches:",
      (destination_master["Accommodation_Count"] == 0).sum())

display(
    destination_master[
        [
            "Location_Name",
            "Mapped_District",
            "Accommodation_Count"
        ]
    ].head(15)
)
```

```python
def get_accommodations_for_destination(destination_name, top_n=10):
    
    # Find the selected destination
    destination = destination_master[
        destination_master["Location_Name"].str.casefold()
        == destination_name.casefold()
    ]

    if destination.empty:
        print("Destination not found.")
        return pd.DataFrame()

    district = destination.iloc[0]["Mapped_District"]

    if pd.isna(district):
        print("District mapping is unavailable for this destination.")
        return pd.DataFrame()

    # Filter accommodations by district
    district_key = str(district).strip().casefold()

    results = accommodation[
        accommodation["District_Key"] == district_key
    ].copy()

    if results.empty:
        print("No accommodation records found for this district.")
        return pd.DataFrame()

    # Display useful accommodation information
    columns_to_show = [
        "Name",
        "Type",
        "Address",
        "Rooms",
        "Grade",
        "District"
    ]

    return results[columns_to_show].head(top_n).reset_index(drop=True)


# Test the function
bentota_accommodations = get_accommodations_for_destination(
    "Bentota Beach",
    top_n=10
)

display(bentota_accommodations)
```

```python
def get_accommodations_for_destination(destination_name, top_n=10):
    
    destination = destination_master[
        destination_master["Location_Name"].str.casefold()
        == destination_name.casefold()
    ]

    if destination.empty:
        print("Destination not found.")
        return pd.DataFrame()

    district = destination.iloc[0]["Mapped_District"]
    city = destination.iloc[0]["Located_City"]

    # District-level candidate list
    results = accommodation[
        accommodation["District_Key"]
        == str(district).strip().casefold()
    ].copy()

    # Search city/destination terms in address and name
    search_terms = {
        str(destination_name).strip().casefold(),
        str(city).strip().casefold()
    }
    search_terms.discard("")

    text = (
        results["Name"].fillna("").astype(str) + " " +
        results["Address"].fillna("").astype(str)
    ).str.casefold()

    results["Location_Text_Match"] = text.apply(
        lambda value: any(term in value for term in search_terms)
    )

    # Show local text matches first
    results = results.sort_values(
        "Location_Text_Match",
        ascending=False
    )

    columns = [
        "Name", "Type", "Address",
        "Rooms", "Grade", "District",
        "Location_Text_Match"
    ]

    return results[columns].head(top_n).reset_index(drop=True)


display(
    get_accommodations_for_destination(
        "Bentota Beach",
        top_n=10
    )
)
```

```python
def recommend_accommodations(
    destination_name,
    preferred_type=None,
    top_n=10
):
    results = get_accommodations_for_destination(
        destination_name,
        top_n=1000
    )

    if results.empty:
        return results

    # Keep only candidates with a text match
    results = results[
        results["Location_Text_Match"]
    ].copy()

    if preferred_type:
        type_matches = results[
            results["Type"].str.casefold()
            == preferred_type.casefold()
        ]

        # Use the preferred type if matches exist
        if not type_matches.empty:
            results = type_matches

    # Sort by room count (descending) as a transparent baseline
    results["Rooms"] = pd.to_numeric(
        results["Rooms"], errors="coerce"
    )

    results = results.sort_values(
        by="Rooms",
        ascending=False,
        na_position="last"
    )

    return results.head(top_n).reset_index(drop=True)
```
