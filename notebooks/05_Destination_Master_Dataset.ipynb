```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

reviews = pd.read_csv(
    "../data/processed/reviews_cleaned.csv"
)

activities = pd.read_csv(
    "../data/processed/activities_cleaned.csv"
)

accommodation = pd.read_csv(
    "../data/processed/accommodation_cleaned.csv"
)

water_sports = pd.read_csv(
    "../data/processed/water_sports_cleaned.csv"
)
```

```python
destination_master = (
    reviews
    .groupby(["Location_Name", "Located_City", "Location_Type"])
    .agg(
        Average_Rating=("Rating", "mean"),
        Review_Count=("Rating", "count")
    )
    .reset_index()
)
```

```python
destination_master.head()
```

```python
destination_master["Average_Rating"] = (
    destination_master["Average_Rating"].round(2)
)
```

```python
most_common_month = (
    reviews.groupby("Location_Name")["Travel_Month"]
    .agg(lambda x: x.mode().iloc[0] if not x.mode().empty else np.nan)
    .reset_index()
)

most_common_month.columns = [
    "Location_Name",
    "Most_Common_Travel_Month"
]
```

```python
destination_master = destination_master.merge(
    most_common_month,
    on="Location_Name",
    how="left"
)
```

```python
print("Number of destinations:", destination_master["Location_Name"].nunique())

print("\nShape:", destination_master.shape)

destination_master.head(10)
```
