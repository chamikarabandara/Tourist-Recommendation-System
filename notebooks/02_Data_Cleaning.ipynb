```python
import pandas as pd
import numpy as np
```

```python
reviews = pd.read_csv(
    "../data/raw/Reviews.csv",
    encoding="cp1252"
)

accommodation = pd.read_csv(
    "../data/raw/Information for Accommodation.csv",
    encoding="cp1252"
)

activities = pd.read_csv(
    "../data/raw/Places for Travel-Dining-Recreational activities and Information of travel agents.csv",
    encoding="cp1252"
)

water_sports = pd.read_excel(
    "../data/raw/Water Sports Centers.xlsx"
)

print("All datasets loaded successfully!")
```

```python
reviews_clean = reviews.copy()
accommodation_clean = accommodation.copy()
activities_clean = activities.copy()
water_sports_clean = water_sports.copy()
```

```python
reviews_clean.columns = reviews_clean.columns.str.strip()
accommodation_clean.columns = accommodation_clean.columns.str.strip()
activities_clean.columns = activities_clean.columns.str.strip()
water_sports_clean.columns = water_sports_clean.columns.str.strip()
```

```python
print(reviews_clean.columns.tolist())
```

```python
reviews_clean["Travel_Date"] = pd.to_datetime(
    reviews_clean["Travel_Date"],
    format="%Y-%m",
    errors="coerce"
)

print(reviews_clean["Travel_Date"].dtype)
print(reviews_clean["Travel_Date"].head())
```

```python
reviews_clean["Travel_Year"] = reviews_clean["Travel_Date"].dt.year
reviews_clean["Travel_Month"] = reviews_clean["Travel_Date"].dt.month

reviews_clean[
    ["Travel_Date", "Travel_Year", "Travel_Month"]
].head()
```

```python
print(reviews_clean["Rating"].min())
print(reviews_clean["Rating"].max())

invalid_ratings = reviews_clean[
    ~reviews_clean["Rating"].between(1, 5)
]

print("Invalid ratings:", len(invalid_ratings))
```

```python
text_columns = [
    "Location_Name",
    "Located_City",
    "Location",
    "Location_Type",
    "User_Location",
    "User_Locale",
    "Title"
]

for col in text_columns:
    reviews_clean[col] = (
        reviews_clean[col]
        .astype(str)
        .str.strip()
    )
```

```python
print(reviews_clean.isnull().sum())

print(
    "Invalid Travel Dates:",
    reviews_clean["Travel_Date"].isna().sum()
)
```

```python
print(
    "Duplicate rows:",
    reviews_clean.duplicated().sum()
)
```

```python
reviews_clean.to_csv(
    "../data/processed/reviews_cleaned.csv",
    index=False,
    encoding="utf-8-sig"
)

print("Cleaned Reviews dataset saved successfully!")
```

Check Accommodation Missing Values

```python
accommodation_clean.isnull().sum()
```

```python
accommodation_text_columns = [
    "Type",
    "Name",
    "Address",
    "Grade",
    "District",
    "AGA Division",
    "PS/MC/UC"
]

for col in accommodation_text_columns:
    accommodation_clean[col] = (
        accommodation_clean[col]
        .astype("string")
        .str.strip()
    )
```

```python
accommodation_clean["Grade"] = (
    accommodation_clean["Grade"]
    .str.upper()
    .str.strip()
)
```

```python
accommodation_clean["Grade"] = (
    accommodation_clean["Grade"]
    .fillna("UNKNOWN")
)
```

```python
print(
    accommodation_clean["Grade"].value_counts()
)
```

```python
accommodation_clean["AGA Division"] = (
    accommodation_clean["AGA Division"]
    .fillna("UNKNOWN")
)
```

```python
accommodation_clean["PS/MC/UC"] = (
    accommodation_clean["PS/MC/UC"]
    .fillna("UNKNOWN")
)
```

```python
print("Missing Longitude:",
      accommodation_clean["Logitiute"].isnull().sum())

print("Missing Latitude:",
      accommodation_clean["Latitude"].isnull().sum())

accommodation_clean["Rooms"] = pd.to_numeric(
    accommodation_clean["Rooms"],
    errors="coerce"
)

print(
    "Missing Rooms:",
    accommodation_clean["Rooms"].isnull().sum()
)
```

```python
print(accommodation_clean.isnull().sum())
```

```python
accommodation_clean.to_csv(
    "../data/processed/accommodation_cleaned.csv",
    index=False,
    encoding="utf-8-sig"
)

print("Cleaned Accommodation dataset saved successfully!")
```
