```python
import pandas as pd
import numpy as np

reviews = pd.read_csv(
    "../data/raw/Reviews.csv",
    encoding="cp1252"
)

accommodation = pd.read_csv(
    "../data/raw/Information for Accommodation.csv",
    encoding="cp1252"
)

activities = pd.read_csv(
    "../data/raw/Places for Travel-Dining-Recreational activities and Information of travel agents.csv",
    encoding="cp1252"
)

water_sports = pd.read_excel(
    "../data/raw/Water Sports Centers.xlsx"
)

print("All datasets loaded successfully!")
```

```python
print("Reviews:", reviews.shape)
print("Accommodation:", accommodation.shape)
print("Activities:", activities.shape)
print("Water Sports:", water_sports.shape)
```

```python
print("========== REVIEWS ==========")
print(reviews.columns.tolist())

print("\n========== ACCOMMODATION ==========")
print(accommodation.columns.tolist())

print("\n========== ACTIVITIES ==========")
print(activities.columns.tolist())

print("\n========== WATER SPORTS ==========")
print(water_sports.columns.tolist())
```

```python
print("========== REVIEWS ==========")
display(reviews.head())

print("========== ACCOMMODATION ==========")
display(accommodation.head())

print("========== ACTIVITIES ==========")
display(activities.head())

print("========== WATER SPORTS ==========")
display(water_sports.head())
```

```python
print("========== REVIEWS ==========")
reviews.info()

print("\n========== ACCOMMODATION ==========")
accommodation.info()

print("\n========== ACTIVITIES ==========")
activities.info()

print("\n========== WATER SPORTS ==========")
water_sports.info()
```

```python
print("========== REVIEWS ==========")
print(reviews.isnull().sum())

print("\n========== ACCOMMODATION ==========")
print(accommodation.isnull().sum())

print("\n========== ACTIVITIES ==========")
print(activities.isnull().sum())

print("\n========== WATER SPORTS ==========")
print(water_sports.isnull().sum())
```

```python
datasets = {
    "Reviews": reviews,
    "Accommodation": accommodation,
    "Activities": activities,
    "Water Sports": water_sports
}

for name, df in datasets.items():
    print(f"\n===== {name} =====")
    missing = df.isnull().sum()
    print(missing[missing > 0])
```

```python
for name, df in datasets.items():
    print(f"{name}: {df.duplicated().sum()} duplicate rows")
```

```python
print("Rating values:")
print(sorted(reviews["Rating"].unique()))

print("\nRating counts:")
print(reviews["Rating"].value_counts().sort_index())
```

```python
print(reviews["Travel_Date"].head())

print("\nData type:")
print(reviews["Travel_Date"].dtype)
```

```python
print("===== Location Types =====")
print(reviews["Location_Type"].value_counts())

print("\n===== Number of Destinations =====")
print(reviews["Location_Name"].nunique())

print("\n===== Number of Cities =====")
print(reviews["Located_City"].nunique())

print("\n===== Number of Travel Months =====")
print(reviews["Travel_Date"].nunique())

print("\n===== Travel Date Range =====")
print(reviews["Travel_Date"].min())
print(reviews["Travel_Date"].max())
```

```python
print("===== Accommodation Types =====")
print(accommodation["Type"].value_counts())

print("\n===== Accommodation Districts =====")
print(accommodation["District"].nunique())

print("\n===== Accommodation Grades =====")
print(accommodation["Grade"].value_counts(dropna=False))
```
