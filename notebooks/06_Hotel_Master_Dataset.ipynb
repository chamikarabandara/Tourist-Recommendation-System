```python
import pandas as pd
import numpy as np
```

```python
accommodation = pd.read_csv(
    "../data/processed/accommodation_cleaned.csv"
)

print("Dataset shape:", accommodation.shape)
print("\nColumns:")
print(accommodation.columns.tolist())

display(accommodation.head())
```

```python
print("Missing values:")
print(accommodation.isnull().sum())

print("\nDuplicate rows:", accommodation.duplicated().sum())

print("\nAccommodation types:")
print(accommodation["Type"].value_counts())

print("\nDistricts:", accommodation["District"].nunique())
```

```python
# Create a separate master dataset
hotel_master = accommodation.copy()

# Clean text fields
text_columns = [
    "Type",
    "Name",
    "Address",
    "Grade",
    "District",
    "AGA Division",
    "PS/MC/UC"
]

for col in text_columns:
    hotel_master[col] = (
        hotel_master[col]
        .astype("string")
        .str.strip()
    )

# Normalize district names for matching
hotel_master["District_Key"] = (
    hotel_master["District"]
    .str.casefold()
    .str.strip()
)

# Ensure room counts are numeric
hotel_master["Rooms"] = pd.to_numeric(
    hotel_master["Rooms"],
    errors="coerce"
)

# Validate
print("Hotel master shape:", hotel_master.shape)
print("Duplicate rows:", hotel_master.duplicated().sum())
print("\nMissing values:")
print(hotel_master.isnull().sum())

display(hotel_master.head())
```

```python
hotel_master.to_csv(
    "../data/processed/hotel_master.csv",
    index=False,
    encoding="utf-8-sig"
)

print("Hotel master dataset saved successfully!")
print("Total accommodation records:", len(hotel_master))
```
